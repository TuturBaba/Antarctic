```python
import sys
import os
import platform
import datetime
import pandas as pd
import traceback
import urllib.request

# Start recording output
log_file = open("test_python.txt", "w", encoding="utf-8")
sys.stdout = log_file
sys.stderr = log_file

print(f"[{datetime.datetime.now()}] === TEST OF PYTHON ENVIRONMENT IN DOCKER ===\n")

# Display Python version
print(f"- Installed Python version: {platform.python_version()}")

print("-------------------------------------------------------------------------")

# Basic math operations test
def test_operations():
    addition = 5 + 10
    multiplication = 7 * 8
    exponential = 2 ** 10
    print(f"- Addition (5 + 10) = {addition}")
    print(f"- Multiplication (7 * 8) = {multiplication}")
    print(f"- Exponential (2^10) = {exponential}")

test_operations()

print("-------------------------------------------------------------------------")

# Create and inspect a DataFrame
df = pd.DataFrame({
    "id": range(1, 6),
    "value": [12.5, 9.8, 15.2, 7.6, 11.3],
    "category": ["A", "B", "A", "C", "B"]
})
print(f"- Created a dataframe with {df.shape[0]} rows and {df.shape[1]} columns")
print("- Dataframe preview:")
print(df.head())
print(f"- Average of 'value' column: {df['value'].mean()}")


print("-------------------------------------------------------------------------")

os.system("conda install matplotlib")
import matplotlib.pyplot as plt
# Package test (example: rasterio as an alt to terra)

os.system("conda install rasterio")
try:
    import rasterio
    print("- Package 'rasterio' loaded successfully")
    print(f"- Package version: {rasterio.__version__}")
except ImportError:
    print("- Package 'rasterio' not found.")
except Exception as e:
    print(f"- ERROR with package loading: {e}")
    traceback.print_exc()

    
print("-------------------------------------------------------------------------")

# Plot test
try:
    plt.figure()
    plt.plot(range(1, 11), [x**2 for x in range(1, 11)], marker='o')
    plt.title("Graphic Test")
    plt.savefig("plot_test.png")
    plt.close()
    print("- Graphic test successful ✓ (saved as plot_test.png)")
except Exception as e:
    print(f"- ERROR during graphic creation: {e}")
    traceback.print_exc()



print("-------------------------------------------------------------------------")

# File system operations test
try:
    current_dir = os.getcwd()
    files = os.listdir(current_dir)
    print(f"- Current working directory: {current_dir}")
    print(f"- Number of files in the directory: {len(files)}")
    if files:
        print(f"- First files found: {', '.join(files[:3])}...")
    
    test_file = "test_python_docker.txt"
    with open(test_file, "w") as f:
        f.write("Test writing to a file\n")

    if os.path.exists(test_file):
        with open(test_file, "r") as f:
            content = f.read().strip()
        print(f"- Content read from the file: {content}")
        os.remove(test_file)
        print("- File deleted")
except Exception as e:
    print(f"- ERROR accessing the file system: {e}")
    traceback.print_exc()

print("-------------------------------------------------------------------------")

# Raster download and read test (mimicking terra raster)
try:
    url = "https://gis.ccamlr.org/geoserver/www/GEBCO2024_5000.tif"
    dest = "GEBCO2024_5000.tif"
    urllib.request.urlretrieve(url, dest)
    print(f"- Download of {dest} completed")

    with rasterio.open(dest) as src:
        print(f"- Raster CRS: {src.crs}")
        print(f"- Raster dimensions: {src.width}x{src.height}")
        # Optionally plot
        plt.imshow(src.read(1), cmap='terrain')
        plt.title("Raster Preview")
        plt.colorbar()
        plt.savefig("raster_preview.png")
        plt.close()
        print("- Raster preview saved as raster_preview.png")
except Exception as e:
    print(f"- ERROR during raster download or plot: {e}")
    traceback.print_exc()

print(f"[{datetime.datetime.now()}] === END ===\n")

# End logging
log_file.close()
sys.stdout = sys.__stdout__
sys.stderr = sys.__stderr__

```
