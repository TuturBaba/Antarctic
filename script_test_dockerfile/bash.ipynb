```bash
echo "[ $(date) ] === TEST OF BASH ENVIRONMENT IN DOCKER ===" > test_bash.txt
echo "-------------------------------------------------------------------------" >> test_bash.txt

# Display Bash version
bash --version >> test_bash.txt

echo "-------------------------------------------------------------------------" >> test_bash.txt

# Basic math operations
add=$((5 + 10))
mul=$((7 * 8))
exp=$((2 ** 10))
echo "- Addition (5 + 10) = $add" >> test_bash.txt
echo "- Multiplication (7 * 8) = $mul" >> test_bash.txt
echo "- Exponential (2^10) = $exp" >> test_bash.txt

echo "-------------------------------------------------------------------------" >> test_bash.txt

# Create and inspect a CSV file
echo "id,value,category" > sample.csv
echo "1,12.5,A" >> sample.csv
echo "2,9.8,B" >> sample.csv
echo "3,15.2,A" >> sample.csv
echo "4,7.6,C" >> sample.csv
echo "5,11.3,B" >> sample.csv

echo "- Created sample.csv" >> test_bash.txt
echo "- First few lines of sample.csv:" >> test_bash.txt
head sample.csv >> test_bash.txt

echo "-------------------------------------------------------------------------" >> test_bash.txt

conda install -y r-terra

if [ $? -eq 0 ]; then
    echo "- Package 'r-terra' installed successfully." >> test_bash.txt
else
    echo "- ERROR: Installation of 'r-terra' failed." >> test_bash.txt
fi

echo "-------------------------------------------------------------------------" >> test_bash.txt

# File system operations
current_dir=$(pwd)
file_count=$(ls | wc -l)
echo "- Current working directory: $current_dir" >> test_bash.txt
echo "- Number of files in the directory: $file_count" >> test_bash.txt
echo "- First few files:" >> test_bash.txt
ls | head -n 3 >> test_bash.txt

test_file="test_bash_file.txt"
echo "Hello from Bash!" > "$test_file"
if [ -f "$test_file" ]; then
    content=$(cat "$test_file")
    echo "- Created file $test_file with content: $content" >> test_bash.txt
    rm "$test_file"
    echo "- File deleted" >> test_bash.txt
fi
echo "-------------------------------------------------------------------------" >> test_bash.txt

output="GEBCO2024_5000_bash.tif"
wget "https://gis.ccamlr.org/geoserver/www/GEBCO2024_5000.tif" -O $output

# Vérifier si le téléchargement a réussi
if [ $? -eq 0 ]; then
    echo "- Download of $output completed successfully." >> test_bash.txt
else
    echo "- ERROR: Download of $output failed." >> test_bash.txt
fi

echo "[ $(date) ] === END ===" >> test_bash.txt
```
