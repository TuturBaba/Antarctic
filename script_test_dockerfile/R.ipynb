```r
# Start recording in a text file
sink("test_R.txt")

cat("[", Sys.time(), "] === TEST OF R ENVIRONMENT IN DOCKER ===\n\n")

# Display R version
r_version <- R.version.string
cat("- Installed R version:", r_version, "\n")

cat("-------------------------------------------------------------------------\n")

# Basic math operations test
test_operations <- function() {
  addition <- 5 + 10
  multiplication <- 7 * 8
  exponential <- 2^10
  cat("- Addition (5 + 10) =", addition, "\n")
  cat("- Multiplication (7 * 8) =", multiplication, "\n")
  cat("- Exponential (2^10) =", exponential, "\n")
}
test_operations()

cat("-------------------------------------------------------------------------\n")

# Create and inspect a dataframe
df <- data.frame(
  id = 1:5,
  value = c(12.5, 9.8, 15.2, 7.6, 11.3),
  category = c("A", "B", "A", "C", "B")
)
cat("- Created a dataframe with", nrow(df), "rows and", ncol(df), "columns\n")
cat("- Dataframe preview:\n")
print(head(df))
cat("- Average of 'value' column:", mean(df$value), "\n")

cat("-------------------------------------------------------------------------\n")

# Plot test using a null PDF device
tryCatch({
  pdf(NULL)
  plot(1:10, (1:10)^2, main = "Graphic Test")
  dev.off()
  cat("- Graphic test successful ✓\n\n")
}, error = function(e) {
  cat("- ERROR during graphic creation:", e$message, "\n\n")
})

cat("-------------------------------------------------------------------------\n")

# Conditional package installation and loading
tryCatch({
  if (!requireNamespace("terra", quietly = TRUE)) {
    cat("- Package 'terra' not found. Installing via conda...\n")
    system("conda install -y r-terra")
  }
  library(terra)
  cat("- Package 'terra' loaded successfully\n")
  cat("- Package version:", as.character(packageVersion("terra")), "\n")
}, error = function(e) {
  cat("- ERROR during package installation or loading:", e$message, "\n\n")
})

cat("-------------------------------------------------------------------------\n")

# File system operations test
current_dir <- getwd()
cat("- Current working directory:", current_dir, "\n")
files <- list.files()
cat("- Number of files in the directory:", length(files), "\n")
if (length(files) > 0) {
  cat("- First files found:", paste(head(files, 3), collapse = ", "), "...\n")
}

test_file <- "test_r_docker.txt"
tryCatch({
  cat("Test writing to a file\n", file = test_file)
  if (file.exists(test_file)) {
    content <- readLines(test_file)
    cat("- Content read from the file:", content, "\n")
    file.remove(test_file)
    cat("- File deleted\n")
  }
}, error = function(e) {
  cat("- ERROR accessing the file system:", e$message, "\n\n")
})

cat("-------------------------------------------------------------------------\n")

# Raster download and plot test
options(timeout = 3600)
download.file("https://gis.ccamlr.org/geoserver/www/GEBCO2024_5000.tif", "GEBCO2024_5000.tif", mode = "wb")
cat("- Download of GEBCO2024_5000.tif completed\n")
plot(rast("GEBCO2024_5000.tif"))

cat("[", Sys.time(), "] === END ===\n\n")

# End recording
sink()
```
